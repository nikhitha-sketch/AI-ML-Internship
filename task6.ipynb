```python
import matplotlib.pyplot as plt

# Sleep Hours Data
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sleep_hours = [7, 6, 8, 5, 7.5, 9, 8.5]

# Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(days, sleep_hours, color="purple", marker="*", s=120)
plt.title("Sleep Hours Throughout the Week")
plt.xlabel("Days")
plt.ylabel("Sleep Hours")
plt.grid(True)
plt.show()

# Bar Chart
plt.figure(figsize=(6,4))
plt.bar(days, sleep_hours, color="skyblue")
plt.title("Daily Sleep Hours")
plt.xlabel("Days")
plt.ylabel("Sleep Hours")
plt.show()

# Line Chart
plt.figure(figsize=(6,4))
plt.plot(days, sleep_hours, marker="o", linestyle="--", color="green", linewidth=2)
plt.title("Weekly Sleep Pattern")
plt.xlabel("Days")
plt.ylabel("Sleep Hours")
plt.grid(True)
plt.show()
```
