```python
# Simple Prediction App: Study Hours → Predicted Score
def predict_score(hours):
    # Basic formula: 5 points per study hour
    score = hours * 5
    # Cap the score at 100
    if score > 100:
        score = 100
    return score
print("Study Hours Prediction App")
try:
    hours = float(input("Enter the number of study hours: "))
    predicted = predict_score(hours)
    print(f"Predicted Score: {predicted}")
except ValueError:
    print("Please enter a valid number.")
```
